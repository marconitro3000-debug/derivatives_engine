## 1. Black-Scholes Model

Under the risk-neutral measure, the underlying follows geometric Brownian motion:

$$dS_t = r\,S_t\,dt + \sigma\,S_t\,dW_t$$

Black & Scholes (1973) derived a closed-form price for European options by constructing a riskless hedging portfolio. The result for a call option is:

$$C = S_0\,N(d_1) - K e^{-rT} N(d_2)$$

$$d_1 = \frac{\ln(S/K) + (r + \sigma^2/2)\,T}{\sigma\sqrt{T}}, \qquad d_2 = d_1 - \sigma\sqrt{T}$$

Put-call parity gives the put price: $P = C - S_0 + K e^{-rT}$.

### Greeks

The sensitivities of the option price to its parameters:

| Greek | Symbol | Definition | Formula |
|-------|--------|------------|--------|
| Delta | $\Delta$ | $\partial C / \partial S$ | $N(d_1)$ |
| Gamma | $\Gamma$ | $\partial^2 C / \partial S^2$ | $n(d_1)/(S\sigma\sqrt{T})$ |
| Vega  | $\mathcal{V}$ | $\partial C / \partial \sigma$ | $S\,n(d_1)\sqrt{T}$ |
| Theta | $\Theta$ | $\partial C / \partial t$ | $-S n(d_1)\sigma/(2\sqrt{T}) - rKe^{-rT}N(d_2)$ |
| Rho   | $\rho$  | $\partial C / \partial r$ | $KTe^{-rT}N(d_2)$ |

where $n(\cdot)$ is the standard normal PDF.

````python
call = price(S, K, T, r, sigma, 'call')
put  = price(S, K, T, r, sigma, 'put')
g    = greeks(S, K, T, r, sigma)

print(f"Call price : {call:.4f}")
print(f"Put price  : {put:.4f}")
print(f"Put-call parity check: C - P = {call - put:.4f}  vs  S - Ke^(-rT) = {S - K*np.exp(-r*T):.4f}")
print()
print("Greeks:")
for k_name, v in g.items():
    print(f"  {k_name:<15} {v:.5f}")
````

## 2. Implied Volatility

Given an observed market price $C^{mkt}$, the **implied volatility** $\hat\sigma$ is the unique $\sigma$ that solves:

$$C_{BS}(\hat\sigma) = C^{mkt}$$

Since $C_{BS}$ is monotone increasing in $\sigma$ and smooth, Newton-Raphson converges rapidly:

$$\hat\sigma_{n+1} = \hat\sigma_n - \frac{C_{BS}(\hat\sigma_n) - C^{mkt}}{\mathcal{V}(\hat\sigma_n)}$$

**Seed:** Brenner-Subrahmanyam ATM approximation

$$\hat\sigma_0 \approx \sqrt{\frac{2\pi}{T}}\,\frac{C}{S}$$

which is exact when $K = S e^{rT}$ (at-the-money forward). Convergence to $|\hat\sigma - \sigma^{true}| < 10^{-10}$ is achieved in 4-5 iterations.

**Fallback:** Brent's method is used when $\mathcal{V} \approx 0$ (deep ITM or OTM options where vega is near zero and Newton-Raphson stalls).

````python
# Round-trip: sigma -> BS price -> implied_vol -> should recover sigma
mkt_price = price(S, K, T, r, sigma, 'call')    # price at true vol
iv_hat    = implied_vol(S, K, T, r, mkt_price)  # recover vol

print(f"True sigma  : {sigma:.10f}")
print(f"Market price: {mkt_price:.6f}")
print(f"Implied vol : {iv_hat:.10f}")
print(f"Round-trip error: {abs(iv_hat - sigma):.2e}")

# Smile: IV at different strikes
strikes = np.linspace(80, 120, 9)
print("\nVolatility smile (flat vol input -> flat IV output is the sanity check):")
for k_i in strikes:
    px = price(S, k_i, T, r, sigma, 'call')
    iv = implied_vol(S, k_i, T, r, px)
    print(f"  K={k_i:6.1f}  BS={px:.4f}  IV={iv:.4f}")
````

## 3. Monte Carlo Simulation

Monte Carlo prices options by simulating many paths of the risk-neutral process and averaging the discounted payoff:

$$\hat C = e^{-rT}\,\frac{1}{N}\sum_{i=1}^N \text{Payoff}(S_T^{(i)})$$

### GBM path simulation

Euler-Maruyama discretisation over $M$ steps of size $\Delta t = T/M$:

$$S_{t+\Delta t} = S_t \exp\!\left[\left(r - \tfrac{\sigma^2}{2}\right)\Delta t + \sigma\sqrt{\Delta t}\,Z_t\right], \quad Z_t \overset{iid}{\sim} \mathcal{N}(0,1)$$

The log-normal form (rather than Euler on $S$ directly) avoids discretisation bias and ensures $S_t > 0$.

### Variance reduction

**Antithetic variates:** for each draw $Z$ also simulate $-Z$. The payoffs are negatively correlated, so their average has lower variance:

$$\hat C^{AV} = e^{-rT}\,\frac{1}{N}\sum_{i=1}^N \frac{\text{Payoff}(S^{Z_i}) + \text{Payoff}(S^{-Z_i})}{2}$$

**Control variate:** use the European BS price (known analytically) as a control. If $C^{BS}$ is known and $\text{Cov}(\hat C, C^{BS}) > 0$:

$$\hat C^{CV} = \hat C_{MC} - \beta\,(\hat C^{BS}_{MC} - C^{BS})$$

where $\hat\beta = \text{Cov}(\hat C, \hat C^{BS}) / \text{Var}(\hat C^{BS})$ is estimated from the same paths.

````python
bs_ref = price(S, K, T, r, sigma, 'call')

print("European call (MC vs BS):")
res = mc_price(S, K, T, r, sigma, 'european_call', n_sims=200_000, seed=42)
print(f"  MC price   : {res['price']:.4f}  (std_err={res['std_error']:.4f})")
print(f"  BS price   : {bs_ref:.4f}")
print(f"  95% CI     : [{res['conf_95_lo']:.4f}, {res['conf_95_hi']:.4f}]")
print(f"  BS in CI   : {res['conf_95_lo'] <= bs_ref <= res['conf_95_hi']}")

print("\nExotic payoffs (all cheaper than European call â€” no-arb sanity check):")
for ptype in ['asian_call', 'barrier_call', 'lookback_call', 'digital_call']:
    r2 = mc_price(S, K, T, r, sigma, ptype, n_sims=100_000, seed=42)
    print(f"  {ptype:<18} {r2['price']:.4f}  se={r2['std_error']:.4f}")
````

## 4. Binomial Tree (CRR)

Cox, Ross & Rubinstein (1979) approximated the GBM with a recombining binomial lattice. At each step $\Delta t = T/N$:

$$u = e^{\sigma\sqrt{\Delta t}}, \quad d = \frac{1}{u}, \quad p = \frac{e^{r\,\Delta t} - d}{u - d}$$

The tree is priced by backward induction from the terminal payoff. At each interior node the option value is the greater of:
- The continuation value: $e^{-r\Delta t}[p\,V_u + (1-p)\,V_d]$
- The intrinsic value (for American options): $\max(S - K, 0)$ for calls

As $N \to \infty$, the CRR price converges to the BS price at rate $O(1/N)$.

The early-exercise premium for an American put is:

$$\text{EEP} = V^{American} - V^{European} \geq 0$$

This premium is zero for American calls on non-dividend-paying stocks (it is never optimal to early-exercise).

````python
bs_call = price(S, K, T, r, sigma, 'call')
bs_put  = price(S, K, T, r, sigma, 'put')

print("Convergence of CRR tree to Black-Scholes:")
print(f"{'N':>6}  {'EU call':>10}  {'AM call':>10}  {'EU put':>10}  {'AM put':>10}  {'EEP':>8}")
for n in [10, 50, 100, 200, 500, 1000]:
    eu_c = binomial_price(S, K, T, r, sigma, 'call', 'european', n)['price']
    am_c = binomial_price(S, K, T, r, sigma, 'call', 'american', n)['price']
    eu_p = binomial_price(S, K, T, r, sigma, 'put',  'european', n)['price']
    am_p = binomial_price(S, K, T, r, sigma, 'put',  'american', n)['price']
    print(f"{n:>6}  {eu_c:>10.4f}  {am_c:>10.4f}  {eu_p:>10.4f}  {am_p:>10.4f}  {am_p - eu_p:>8.4f}")

print(f"\nBlack-Scholes reference: call={bs_call:.4f}  put={bs_put:.4f}")
print("American call EEP should be 0 (confirmed: no early exercise on non-dividend stock)")
````

## 5. Heston Stochastic Volatility Model

The main limitation of Black-Scholes is the constant-volatility assumption â€” empirically, implied volatility is not flat across strikes (volatility smile/skew). Heston (1993) introduced a mean-reverting stochastic variance process:

$$dS_t = r\,S_t\,dt + \sqrt{v_t}\,S_t\,dW_t^S$$
$$dv_t = \kappa(\theta - v_t)\,dt + \xi\sqrt{v_t}\,dW_t^v, \qquad \langle dW^S, dW^v \rangle = \rho\,dt$$

| Parameter | Interpretation |
|-----------|---------------|
| $v_0$ | Initial instantaneous variance |
| $\kappa > 0$ | Speed of mean reversion |
| $\theta > 0$ | Long-run variance (vol-of-vol anchor) |
| $\xi > 0$ | Volatility of volatility |
| $\rho \in (-1,1)$ | Spot-vol correlation (negative for equity skew) |

**Feller condition:** $2\kappa\theta \geq \xi^2$ guarantees $v_t > 0$ a.s. (the process never reaches zero). Often violated on real calibrated data.

### Characteristic-function pricing

Heston showed the conditional characteristic function of $\ln S_T$ is available in closed form. The call price follows by Gil-Pelaez Fourier inversion:

$$C = S_0 P_1 - K e^{-rT} P_2, \qquad P_j = \frac{1}{2} + \frac{1}{\pi}\int_0^\infty \mathrm{Re}\!\left[\frac{e^{-i\phi \ln K}\,\varphi_j(\phi)}{i\phi}\right]d\phi$$

We use the Albrecher et al. (2007) 'little Heston trap' formulation of $\varphi$ for numerical stability at large $\phi$.

````python
hp = HestonParams(v0=0.04, kappa=2.0, theta=0.04, xi=0.4, rho=-0.6)
print(f"HestonParams: {hp}")
print(f"Feller condition (2*kappa*theta >= xi^2): {hp.feller_satisfied()}")
print(f"  2*kappa*theta = {2*hp.kappa*hp.theta:.4f}  xi^2 = {hp.xi**2:.4f}")

print("\nHeston smile vs Black-Scholes (flat vol = 20%):")
print(f"{'K':>6}  {'Heston':>10}  {'BS flat':>10}  {'Heston IV':>12}")
for k_i in [85, 90, 95, 100, 105, 110, 115]:
    h_px = heston_price(S, k_i, T, r, hp, 'call')
    bs_px = price(S, k_i, T, r, sigma, 'call')
    try:
        h_iv = implied_vol(S, k_i, T, r, h_px)
    except Exception:
        h_iv = float('nan')
    print(f"{k_i:>6}  {h_px:>10.4f}  {bs_px:>10.4f}  {h_iv:>12.4f}")

print("\nNote: Heston IV is not flat â€” the model generates a smile from correlation rho.")
````

## 6. SVI â€” Stochastic Volatility Inspired

Gatheral (2004) proposed a parametric form for the total implied variance $w(k) = \sigma_{IV}^2(k)\,T$ as a function of log-moneyness $k = \ln(K/F)$:

$$w(k) = a + b\left[\rho(k - m) + \sqrt{(k - m)^2 + \sigma^2}\right]$$

| Parameter | Effect on smile |
|-----------|----------------|
| $a \geq 0$ | Overall variance level (vertical shift) |
| $b \geq 0$ | Slope (steepness of wings) |
| $\rho \in (-1,1)$ | Skew â€” tilts the smile left/right |
| $m \in \mathbb{R}$ | Horizontal shift of the smile minimum |
| $\sigma > 0$ | Curvature (smoothness of the ATM region) |

SVI satisfies Lee's moment formula (wings grow at most linearly in $|k|$), which is a necessary condition for absence of calendar-spread arbitrage.

**Butterfly no-arbitrage:** sufficient condition $b(1 + |\rho|) \leq 4$. Necessary and sufficient conditions are more involved; we check the sufficient condition.

Compared to Heston, SVI is:
- Much faster to fit (one unconstrained optimisation per slice vs joint fit over all maturities)
- More flexible (can exactly match any observed smile shape)
- Not a dynamic model â€” no PDE, no path simulation
- Per-maturity only (calendar spread no-arb requires additional constraints across slices)

````python
sp = SVIParams(a=0.04, b=0.40, rho=-0.30, m=0.0, sigma=0.20)
print(f"SVIParams: {sp}")
print(f"Butterfly arb-free: {is_butterfly_arbitrage_free(sp)}")
print(f"  b*(1+|rho|) = {sp.b*(1+abs(sp.rho)):.4f}  (must be <= 4)")

F = S * np.exp(r * T)
k_vals = np.linspace(-0.3, 0.3, 13)
iv_svi = implied_vol_svi(k_vals, T, sp)

print("\nSVI implied vol curve:")
print(f"{'k':>8}  {'K':>8}  {'SVI IV':>10}")
for k_v, iv_v in zip(k_vals, iv_svi):
    print(f"{k_v:>8.3f}  {F*np.exp(k_v):>8.2f}  {iv_v:>10.4f}")
````

## 7. Calibration

Calibration fits model parameters $\theta$ to a set of observed market implied volatilities $\{\sigma_i^{mkt}\}$ at strikes $K_i$ and maturities $T_i$ by minimising the weighted mean-squared error:

$$\min_\theta \sum_{i=1}^N w_i \left(\sigma_i^{model}(\theta) - \sigma_i^{mkt}\right)^2$$

Weights $w_i = \sqrt{\text{volume}_i + 1}$ give more influence to liquid options.

We use `scipy.optimize.least_squares` with the Trust Region Reflective algorithm (TRF), which handles bound constraints efficiently.

### Warm start and EWMA smoothing

Day-to-day, parameters should not jump discontinuously (real vol surfaces move smoothly). Two mechanisms help:

1. **Warm start:** initialise the optimiser from yesterday's fitted parameters instead of a default seed. Converges in fewer iterations and stays in a sensible region.

2. **EWMA blend:** after fitting, blend the new parameters $\theta^{new}$ with the old $\theta^{old}$:

$$\theta^{final} = \alpha\,\theta^{new} + (1-\alpha)\,\theta^{old}, \qquad \alpha \in (0,1]$$

Typical $\alpha = 0.3$: new data has 30% weight; smooths noise without lagging too much.

````python
# Generate a synthetic option chain from known Heston parameters
true_params = HestonParams(v0=0.04, kappa=2.0, theta=0.04, xi=0.4, rho=-0.6)
loader = SyntheticLoader(risk_free_rate=r)
md = loader.load(spot=S, true_params=true_params, noise=0.001, seed=42)

print(f"Synthetic chain: {md.strikes.shape[0]} options, "
      f"{len(np.unique(md.maturities))} maturities, spot={md.spot}")

# Calibrate Heston
result_h = calibrate("heston", md)
print(f"\nHeston calibration: {result_h.summary()}")
print("Fitted vs true parameters:")
true_dict = true_params.to_dict()
for param, fitted_val in result_h.params.items():
    print(f"  {param:<8} fitted={fitted_val:.4f}  true={true_dict[param]:.4f}  "
          f"error={abs(fitted_val - true_dict[param]):.4f}")

# Calibrate SVI
result_s = calibrate("svi", md)
print(f"\nSVI calibration: {result_s.summary()}")
print(f"Fitted {len(result_s.params['slices'])} slices")
````

````python
# Price from the calibrated Heston surface
from core.models.heston import HestonParams

hp_fit = HestonParams.from_dict(result_h.params)
print("Pricing grid from calibrated Heston surface:")
print(f"{'K':>8}  {'T':>6}  {'Heston':>10}  {'BS (true vol)':>14}")
for k_i in [90, 95, 100, 105, 110]:
    for t_i in [0.25, 1.0]:
        h_px  = heston_price(S, k_i, t_i, r, hp_fit, 'call')
        bs_px = price(S, k_i, t_i, r, np.sqrt(true_params.v0), 'call')
        print(f"{k_i:>8}  {t_i:>6.2f}  {h_px:>10.4f}  {bs_px:>14.4f}")
````

## 8. Volatility Surface Interpolation

Given a sparse set of market quotes $\{(K_i, T_j, \hat\sigma_{ij})\}$, we often need to evaluate the IV surface at arbitrary $(K, T)$ points not in the grid â€” for example, to price a bespoke OTC option or to compute sensitivities.

Two non-parametric approaches are implemented:

**Bivariate spline** (`method='spline'`): fits a `RectBivariateSpline` over a rectangular $(K, T)$ grid. Fast and smooth; requires the market quotes to form (or be interpolated to) a regular grid.

**Radial basis functions** (`method='rbf'`): fits an `RBFInterpolator` with thin-plate spline kernel to scattered observations. More flexible; slower for large datasets.

Unlike parametric models (Heston, SVI), these methods make no assumption about the shape of the smile and can exactly fit any observed surface. The tradeoff: no guarantee of no-arbitrage, and extrapolation is unreliable.

````python
# Build a vol surface from synthetic market quotes
strikes_grid    = np.array([85., 90., 95., 100., 105., 110., 115.])
maturities_grid = np.array([0.25, 0.5, 1.0, 2.0])

# Generate IVs from a known Heston model (acts as our 'market')
iv_grid = np.zeros((len(strikes_grid), len(maturities_grid)))
for j, t_j in enumerate(maturities_grid):
    for i, k_i in enumerate(strikes_grid):
        px = heston_price(S, k_i, t_j, r, hp, 'call')
        try:
            iv_grid[i, j] = implied_vol(S, k_i, t_j, r, px)
        except Exception:
            iv_grid[i, j] = np.nan

surf = VolSurface(strikes_grid, maturities_grid, iv_grid, method='spline')

# Query at arbitrary points
print("IV at arbitrary (K, T) via spline surface:")
test_points = [(92.5, 0.375), (100.0, 0.75), (107.5, 1.5)]
for k_q, t_q in test_points:
    iv_q = surf.iv(k_q, t_q)
    h_px = heston_price(S, k_q, t_q, r, hp, 'call')
    h_iv = implied_vol(S, k_q, t_q, r, h_px)
    print(f"  K={k_q:6.1f}  T={t_q:.3f}  surf IV={float(iv_q):.4f}  Heston IV={h_iv:.4f}  "
          f"diff={abs(float(iv_q)-h_iv):.4f}")
````
